```python
import pandas as pd 
import numpy as np 
```

```python
churn=pd.read_csv("CSV DataSets/Churn_Modelling.csv")
churn.info()
```

```python
import seaborn as sns
# import matplotlib.pyplot as plt
from matplotlib import pyplot as plt
```

# Pandas

```python
churn.plot.scatter(x='Age', y='CreditScore')
```

# Matplolib

```python
# Scatter plot
plt.scatter(churn['Age'], churn['CreditScore'])
plt.xlabel('Age')  # Set x-axis label
plt.ylabel('CreditScore')  # Set y-axis label
```

# Seaborn

```python
sns.scatterplot(data=churn, x='Age', y='CreditScore')
```

# Basic Line Plot

```python
x_coords = [2, 3, 5, 7]
y_coords = [4, 6, 8, 10]

plt.plot(x_coords, y_coords) # lineplot
plt.scatter(x_coords, y_coords) # scatterplot

plt.grid()
plt.title("Line Chart")
plt.xlabel('x - values')
plt.ylabel('y - values')
plt.legend(['Line'])
plt.show()
```

# A Stem Plot

```python
x_coords = [2, -3, 5, -7]
y_coords = [4, 4, -8, 10]

plt.stem(x_coords,y_coords)
plt.grid()
plt.title('Stem Plot')
plt.legend(['Stem'])
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.show()
```

# A Bar Plot (plt.figure(figsize=()) exclusively in BarPlots

```python
x_coords = [2, 3, 5, 7]
y_coords = [4, 4, 8, 10]

plt.bar(x_coords,y_coords)
plt.grid()
plt.legend(['Line'])
plt.title('Bar Plot')
plt.xlabel('x - values')
plt.ylabel('y - values')
plt.show()
```

```python
x = ['A','B','C','D','E']
y1 = [55,40,30,60,55]
y2 = [50,30,20,50,60]

plt.figure(figsize=(9,4))

x_axis=np.arange(len(x))
bar_width=0.4

plt.bar(x_axis-bar_width/2, y1, bar_width)
plt.bar(x_axis+bar_width/2, y2, bar_width)

plt.xlabel("x-axis")
plt.ylabel("y-axis")

plt.legend(['Series1','Series2'])
plt.grid()
plt.show()
```

```python
Women = [115, 215, 250, 300]
Men = [114, 230, 400, 370]
x_axis = np.arange(len(Men))
width = 0.25
plt.figure(figsize=(10, 5))
plt.bar(x_axis, Women, color = 'b', width = width, edgecolor = 'black', label='Women')
plt.bar(x_axis + width, Men, color = 'r', width = width, edgecolor = 'g', label='Men')
plt.xlabel("Year")
plt.ylabel("Number of people voted")
plt.title("Number of people voted in each year")
# plt.grid(linestyle='--')
plt.xticks(x_axis + width/2,['2018','2019','2020','2021'])
plt.legend()
plt.grid()
plt.show()
```

# A Scatterplot

```python
x_coordinates=np.random.randn(1,1001)
y_coordinates=np.random.randn(1,1001)
plt.scatter(x_coordinates,y_coordinates)
plt.grid()
plt.title('Scatter')
plt.legend(['points'])
plt.xlabel('X-axis')
plt.ylabel('Y-axis')

plt.show()
```

# A Pie Chart

```python
plt.title('ML Subjects to study:')
data = [12, 25, 50, 36, 19]
activities = ['NLP','Neural Network', 'Data analytics', 'Quantum Computing', 'Machine Learning']
color=['green','purple','orange','red','cyan']
plt.pie(data, labels=activities, colors=color, startangle=90, shadow=True, explode =(0,0,0.1,0,0.2),autopct ='%1.2f%%')
plt.show()
```

# Area PLot

```python
days = [1,2,3,4,5]
age = [63, 81, 52, 22, 37]
weight = [17, 28, 72, 52, 32]
plt.stackplot(days, age, weight, colors=['c','green'])
plt.xlabel('Fluctuation with time')
plt.ylabel('Days')
plt.title('Weather report using Area Plot')
plt.legend(['Weather Predicted', 'Weather Change happened'])
# Print the chart
plt.show()
```

```python
data = np.random.rand(50000)
# Plotting a basic histogram
plt.hist(data, bins=50, color='skyblue', edgecolor='black')
# Adding labels and title
plt.xlabel('Values')
plt.ylabel('Frequency')
plt.title('Basic Histogram')
# Display the plot
plt.show()
```
