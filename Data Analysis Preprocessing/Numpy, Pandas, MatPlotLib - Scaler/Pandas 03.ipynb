# All Methods of Pandas in PDF is covered in this except last few nullValue and duplicates

```python
import pandas as pd
data=[('Nik',34,'2022-22-09'),
      ('Nish',4,'2012-92-09'),
      ('bye',39,'2025-81-09'),
      ('Hello',36,'2021-90-09')]
L=['A','B','C','D']

# Index and column can be added separately
df=pd.DataFrame(data, index=L)
df=pd.DataFrame(data, columns=['Name','Age','Date'])

#  or in one line also
df=pd.DataFrame(data, index=L, columns=['Name','Age','Date'])
print(df)
```

```python
df['Marks']=[99,98,90,97]
print(df)
```

```python
student_marks = [
    {"Math": 85, "Science": 92, "English": 88},
    {"Math": 78, "Science": 80, "English": 75},
    {"Math": 95, "Science": 89, "English": 93},
    {"Math": 70, "Science": 76, "English": 68},
    {"Math": 88, "Science": 84, "English": 90}]
st_name=['Aman','Rahul','Shadab','Asif','Gulab']
df=pd.DataFrame(student_marks,index=st_name)
df
```

# 'loc' and 'iloc' attributes

```python
print(df.loc['Aman'])
print(df.iloc[0])
```

```python
print(df.loc['Aman':'Shadab']) 
print('_________________________________')
print(df.iloc[0:2]) # 0 -> 1
```

```python
print(df.loc[['Aman','Shadab']]) 
print('_________________________________')
print(df.iloc[[0,2]])
```

# Filtering Data 

```python
print(df>=80)
print('_________________________________')
print(df[df>=80])
```

# Accessing a particular index in a particular column in data

```python
print(f'Shadab has secured {df['Science']['Shadab']} in Science')
asif_eng=df['English']['Asif']
print(f'Asif has secured {asif_eng} in {df.columns[2]}')
```

```python
# Selecting all rows of column 1
print(df.iloc[:,0])
# Selecting all columns of row 1
print(df.iloc[0,:])
```

```python
# Transpose
print(df.T)
```

# Sorting by Index and Column
# 0 -> Index
# 1 -> Columns

```python
print('Sorting by Index')
print(df.sort_index(axis=0)) # Sorts order the rows/index (ascending by default)
print(df.sort_index(axis=1)) # Sorts order the columns (ascending by default)

print(df.sort_index(ascending=False)) # Sorts order the rows/index (Descending)
print(df.sort_index(axis=1, ascending=False)) # Sorts order the columns in descending
```
